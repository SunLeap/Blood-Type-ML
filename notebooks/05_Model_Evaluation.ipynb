```python
# CMD / purpose: Run this cell first in each notebook.
# Makes imports work whether Jupyter starts from the project root or notebooks/.
from pathlib import Path
import os, sys
ROOT = Path.cwd()
if not (ROOT / 'src').is_dir():
    ROOT = ROOT.parent
if not (ROOT / 'src').is_dir():
    raise RuntimeError('Open Jupyter from the blood-shortage-ml project directory.')
sys.path.insert(0, str(ROOT))
print('Project root:', ROOT)
```

```python
# CMD / purpose: display saved classifier/regressor validation and held-out test scores.
import pandas as pd
import matplotlib.pyplot as plt
from src.prediction import CLASS_METRICS_PATH, REG_METRICS_PATH, predict_next_week
classification = pd.read_csv(CLASS_METRICS_PATH)
regression = pd.read_csv(REG_METRICS_PATH)
display(classification[['model','selected','validation_macro_f1','test_accuracy','test_macro_f1','test_high_recall']].round(3))
display(regression[['model','selected','validation_mae','test_mae','test_rmse','test_mape_percent']].round(2))
```

```python
# CMD / purpose: visualize classification macro F1 and regression MAE on the test period.
fig, ax = plt.subplots(figsize=(7,4))
classification.set_index('model')['test_macro_f1'].plot(kind='bar',ax=ax)
ax.set_title('Classification held-out test macro F1');ax.set_ylabel('Macro F1');plt.tight_layout();plt.show()
fig, ax = plt.subplots(figsize=(7,4))
regression.set_index('model')['test_mae'].plot(kind='bar',ax=ax)
ax.set_title('Regression held-out test MAE');ax.set_ylabel('Units (lower is better)');plt.tight_layout();plt.show()
```

```python
# CMD / purpose: show transparent next-week predictions from last observed week.
forecast = predict_next_week()
display(forecast[['blood_type','last_observed_week','forecast_week','observed_risk','predicted_risk','risk_confidence','predicted_demand','risk_alert']].round(3))
```
